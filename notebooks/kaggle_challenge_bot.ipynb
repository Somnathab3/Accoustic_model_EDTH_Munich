## 🏃 Run Challenge Bot

This cell runs the challenge bot with:

- HTTP connection pooling

- Automatic model detection

```python
# Run challenge bot with filtered output
import subprocess
import sys
import time
from datetime import datetime

print("="*60)
print("🚀 STARTING CHALLENGE BOT")
print("="*60)
print(f"Started at: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print(f"Max iterations: 100000")
print(f"Delay: 0.0s (fastest)")
print("="*60)
print()

# Start the bot
cmd = [
    sys.executable,
    'sota_challenge_bot.py',
    '--delay', '0.0',
    '--max-iterations', '100000'
]

process = subprocess.Popen(
    cmd,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
    universal_newlines=True
)

# Filter and display output
line_count = 0
show_every_n = 10  # Show every 10th submission
current_iteration = 0

try:
    for line in process.stdout:
        line = line.strip()
        
        # Always show these
        if any(x in line for x in [
            '✓ Using',
            'Initialization complete',
            'CHALLENGE BOT',
            'FINAL SUMMARY',
            'Performance Analysis',
            'Results saved',
            'Stopped by user',
            '❌',  # Errors
            'ERROR',
            'Failed'
        ]):
            print(line)
            continue
        
        # Show iteration results (filtered)
        if line.startswith('[') and ']' in line:
            # Extract iteration number
            try:
                iter_num = int(line.split('[')[1].split(']')[0])
                current_iteration = iter_num
                
                # Show every Nth iteration or every correct answer
                if iter_num % show_every_n == 0 or '✓' in line:
                    print(line)
            except:
                pass
        
        # Show summary lines
        elif line.startswith('─') or line.startswith('Attempts:') or line.startswith('Accuracy:'):
            print(line)
        
        # Show timing info occasionally
        elif '⏱️' in line and current_iteration % 50 == 0:
            print(line)
        
        line_count += 1
        
        # Flush output every 20 lines
        if line_count % 20 == 0:
            sys.stdout.flush()

except KeyboardInterrupt:
    print("\n⏹️  Stopping bot...")
    process.terminate()
    process.wait()
    print("✓ Bot stopped")

# Wait for completion
return_code = process.wait()
print(f"\n✓ Bot finished with return code: {return_code}")
print(f"Ended at: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```

## 📊 View Results

Analyze the challenge results

```python
# Display results summary
import pandas as pd
from pathlib import Path

csv_path = Path('challenge_results/results.csv')

if csv_path.exists():
    df = pd.read_csv(csv_path)
    
    print("="*60)
    print("📊 RESULTS SUMMARY")
    print("="*60)
    
    total = len(df)
    correct = df[df['score_awarded'] > 0].shape[0]
    accuracy = (correct / total * 100) if total > 0 else 0
    
    print(f"\nTotal submissions: {total}")
    print(f"Correct: {correct}")
    print(f"Accuracy: {accuracy:.1f}%")
    
    if total > 0:
        print(f"\nTotal score: {df['total_score'].max()}")
        print(f"Average score per submission: {df['score_awarded'].mean():.1f}")
        print(f"\nAverage inference time: {df['inference_time'].mean():.4f}s")
        print(f"Average total time: {df['total_time'].mean():.2f}s")
        
        # Show last 5 submissions
        print("\nLast 5 submissions:")
        print(df[['timestamp', 'predicted', 'score_awarded', 'total_score', 'total_time']].tail())
        
        # Show high scores
        high_scores = df[df['score_awarded'] >= 190]
        if len(high_scores) > 0:
            print(f"\n🏆 High scores (≥190): {len(high_scores)} submissions")
            print(f"   Average total_time: {high_scores['total_time'].mean():.2f}s")
else:
    print("❌ No results file found yet")
```
